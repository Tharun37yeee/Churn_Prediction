```python
import catboost as cb
import lightgbm as lgb
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score

# -------------------------------------------------------------
# 1. Compute Imbalance Weight (Negatives / Positives)
# -------------------------------------------------------------
# Assuming y_train contains 0s and 1s:
num_neg = np.sum(y_train == 0)
num_pos = np.sum(y_train == 1)
imbalance_ratio = num_neg / num_pos
print(f"Computed scale_pos_weight: {imbalance_ratio:.2f}")


# -------------------------------------------------------------
# 2. LightGBM Implementation
# -------------------------------------------------------------
print("\n--- Training LightGBM ---")

# If you have raw categorical string columns, convert them to pandas 'category':
# cat_cols = ['Contract', 'PaymentMethod', 'InternetService']
# x_train[cat_cols] = x_train[cat_cols].astype('category')
# x_test[cat_cols] = x_test[cat_cols].astype('category')

lgb_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    num_leaves=31,
    max_depth=5,
    scale_pos_weight=imbalance_ratio,  # Balances precision vs recall
    random_state=42,
    verbosity=-1,
)

lgb_model.fit(
    x_train,
    y_train,
    eval_set=[(x_test, y_test)],
    eval_metric="binary_logloss",
    callbacks=[
        lgb.early_stopping(stopping_rounds=20, verbose=False),
        lgb.log_evaluation(period=0),
    ],
)

# Predictions & Probabilities
lgb_probs = lgb_model.predict_proba(x_test)[:, 1]
# Use 0.5 default, or replace with your chosen threshold (e.g., 0.55)
lgb_preds = (lgb_probs >= 0.55).astype(int)


# -------------------------------------------------------------
# 3. CatBoost Implementation
# -------------------------------------------------------------
print("--- Training CatBoost ---")

# CatBoost can take column names directly for categorical features:
# cat_features = ['Contract', 'PaymentMethod', 'InternetService']

cat_model = cb.CatBoostClassifier(
    iterations=300,
    learning_rate=0.05,
    depth=5,
    auto_class_weights="Balanced",  # Automatically calculates and balances classes
    eval_metric="Logloss",
    random_seed=42,
    verbose=False,
    # cat_features=cat_features    # Uncomment if passing raw unencoded string columns
)

cat_model.fit(
    x_train,
    y_train,
    eval_set=(x_test, y_test),
    early_stopping_rounds=20,
    verbose=False,
)

# Predictions & Probabilities
cat_probs = cat_model.predict_proba(x_test)[:, 1]
cat_preds = (cat_probs >= 0.55).astype(int)


# -------------------------------------------------------------
# 4. Compare Performance
# -------------------------------------------------------------
def evaluate_model(name, y_true, y_pred, y_prob):
    print(f"\n[{name} Results]")
    print(f"Accuracy : {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"Recall   : {recall_score(y_true, y_pred):.4f}")
    print(f"ROC-AUC  : {roc_auc_score(y_true, y_prob):.4f}")


evaluate_model("LightGBM", y_test, lgb_preds, lgb_probs)
evaluate_model("CatBoost", y_test, cat_preds, cat_probs)
```

```python
# Average the predicted probabilities
ensemble_probs = (lgb_probs + cat_probs) / 2

# Evaluate at threshold 0.50 or 0.52
ensemble_preds = (ensemble_probs >= 0.52).astype(int)

print("Ensemble ROC-AUC:", roc_auc_score(y_test, ensemble_probs))
print("Ensemble Recall :", recall_score(y_test, ensemble_preds))
print("Ensemble Precision:", precision_score(y_test, ensemble_preds))
```
